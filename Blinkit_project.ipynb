```python
#Find Out What Is Driving Better Sales Performance
```

```python
sale_item=df.groupby('Item Type')['Sales'].sum().sort_values(ascending=False)
print(sale_item)
```

```python
sns.barplot(x=sale_item.index, y=sale_item.values)
plt.xticks(rotation=45)
plt.title("Sales by Product Type")
plt.show()
```

```python
outlet_sales = df.groupby("Outlet Type")["Sales"].sum().sort_values(ascending=False)
print("Top Outlet Types by Sales:\n", outlet_sales)
```

```python
df.columns
```

```python
sns.barplot(x=outlet_sales.index, y=outlet_sales.values)
plt.title("Sales by Outlet Type")
plt.show()
```

```python
corr_features = ["Item Visibility", "Item Weight", "Rating", "Sales"]
sns.pairplot(df[corr_features])
plt.suptitle("Feature Relationships with Sales", y=1.02)
plt.show()
```

```python
top_products = df.groupby("Item Type")["Sales"].mean().sort_values(ascending=False).head(5)
low_products = df.groupby("Item Type")["Sales"].mean().sort_values().head(5)
```

```python
print("Top Performing Products:\n", top_products)
print("Low Performing Products:\n", low_products)
```

```python
#Help Management Make Better Business Decisions
```

```python
top_contributors = df.groupby("Item Type")["Sales"].sum().sort_values(ascending=False).head(10)
print("Top 10 Products by Sales Contribution:\n", top_contributors)
```

```python
outlet_location_sales = df.groupby(["Outlet Location Type", "Item Type"])["Sales"].sum().sort_values(ascending=False)
print("Outlet & Location Performance:\n", outlet_location_sales)
```

```python
avg_sales_by_outlet = df.groupby("Item Type")["Sales"].mean()
print("Average Sales by Outlet Type:\n", avg_sales_by_outlet)
```

```python
underperforming_outlets = avg_sales_by_outlet[avg_sales_by_outlet < avg_sales_by_outlet.mean()]
print("Underperforming Outlets:\n", underperforming_outlets)
```

```python
print("\n--- Practical Suggestions ---")
print("1. Focus marketing on top-selling product categories.")
print("2. Improve visibility and ratings of low-performing products.")
print("3. Strengthen outlets in underperforming locations.")
print("4. Optimize inventory for high-demand products.")
```
